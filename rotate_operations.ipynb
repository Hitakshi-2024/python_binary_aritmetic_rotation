```python
# 8-bit Rotate Left (ROL) and Rotate through Carry Left (RCL)

def rol(binary, positions):
    value = int(binary, 2)
    print("\nROL - Rotate Left Without Carry")
    print("Original number:", binary)

    for step in range(1, positions + 1):
        msb = (value >> 7) & 1
        value = ((value << 1) & 0xFF) | msb
        result = format(value, '08b')
        print(f"Step {step}: {result}")

    print("Final ROL result:", format(value, '08b'))


def rcl(binary, positions, carry):
    value = int(binary, 2)
    print("\nRCL - Rotate Left Through Carry")
    print("Original number:", binary)
    print("Initial carry flag:", carry)

    for step in range(1, positions + 1):
        old_msb = (value >> 7) & 1
        value = ((value << 1) & 0xFF) | carry
        carry = old_msb

        result = format(value, '08b')
        print(f"Step {step}: {result}, Carry = {carry}")

    print("Final RCL result:", format(value, '08b'))
    print("Final carry flag:", carry)


number = "10110110"
positions = 2

rol(number, positions)

# Assume the initial carry flag is 0 for RCL.
rcl(number, positions, carry=0)
```
