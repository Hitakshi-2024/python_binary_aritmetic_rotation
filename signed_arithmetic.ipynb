```python
# Signed Binary Arithmetic using 8-bit 2's Complement

BITS = 8
MIN_VALUE = -128
MAX_VALUE = 127


def to_binary(number):
    """Convert a signed integer to 8-bit 2's complement."""
    return format(number & 0xFF, '08b')


def from_binary(binary):
    """Convert 8-bit 2's complement binary to signed decimal."""
    value = int(binary, 2)
    if value & 0x80:
        value -= 256
    return value


def subtract(a, b, expression):
    print("\nExpression:", expression)
    print("-" * 45)

    # Subtraction is performed by adding the 2's complement
    # (bitwise inverse plus one) of the second operand.
    a_bin = to_binary(a)
    b_bin = to_binary(b)
    negative_b_bin = to_binary(-b)

    print(f"First operand  ({a:+d}): {a_bin}")
    print(f"Second operand ({b:+d}): {b_bin}")
    print(f"2's complement of second operand: {negative_b_bin}")

    result_binary_value = (int(a_bin, 2) + int(negative_b_bin, 2))
    result_8bit = format(result_binary_value & 0xFF, '08b')
    result_decimal = from_binary(result_8bit)

    # Signed overflow occurs when the mathematical result
    # is outside the range -128 to +127.
    actual_result = a - b
    overflow = not (MIN_VALUE <= actual_result <= MAX_VALUE)

    print("\nBinary addition:")
    print(f"  {a_bin}")
    print(f"+ {negative_b_bin}")
    print("  --------")
    print(f"  {result_8bit}")

    print("\n8-bit binary result:", result_8bit)
    print("8-bit signed decimal result:", result_decimal)
    print("Mathematical result:", actual_result)
    print("Arithmetic overflow:", "YES" if overflow else "NO")


# Verify both expressions
subtract(45, 19, "45 - (+19)")
subtract(-35, -12, "(-35) - (-12)")
```
